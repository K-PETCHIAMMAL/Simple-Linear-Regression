```python
import pandas as pd
```

```python
#Placing Our input under a variqable name "dataset"
```

```python
dataset=pd.read_csv("Salary_Data.csv")
```

```python
#Calling to check whether our input data are memorized under the given variable dataset.
```

```python
dataset
```

```python
#Variable "independent" contains only the input - Years of Experience(Separated from Salary_Data.csv)
```

```python
independent=dataset[["YearsExperience"]]
```

```python
#Calling to verify the contents of the variable "independent"
```

```python
independent
```

```python
# Saving the salary data from the given input under the variable "dependent" 
```

```python
dependent=dataset[["Salary"]]
```

```python
# Calling to check the contents of variable dependent
```

```python
dependent
```

```python
# Splitting of test and training dataset with the help of sklearn library
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.30,random_state=0)
```

```python
#Learning phase
```

```python
#importing the LinearRegression class from scikit-learn’s linear_model module.
#this is the machine learning model that will learn the relationship between input (X) and output (Y).
#creating an object (or instance) of the LinearRegression class, called regressor.
#This object is like a blank model — it hasn't learned anything yet.
#regressor.fit=This line trains the model (i.e., fits it to the data).
#It finds the best line that matches the pattern in your data using the training set (x_train and y_train). 
#Also after this line has been executed it is meant that "weight and Bias " had been calculated.
```

```python
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
regressor.fit(x_train,y_train)
```

```python
LinearRegression()
```

```python
#Bias and weight are stored under regressor variable
```

```python
weight=regressor.coef_
```

```python
weight
```

```python
bias=regressor.intercept_
```

```python
bias
```

```python
#Make Predictions (on new or test data)
```

```python
y_pred=regressor.predict(x_test)
```

```python
#Evaluate the Model’s Accuracy
```

```python
# MSE (Mean Squared Error): Lower is better.
# R² Score: Tells how well your model fits the data. 1 means perfect fit, 0 means no fit.
```

```python
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
```

```python
r_score
```
